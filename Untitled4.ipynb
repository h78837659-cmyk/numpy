```python
def categorize_col(df, col, labels):

    edges = [
        df[col].min(),
        df[col].quantile(0.25),
        df[col].median(),
        df[col].quantile(0.75),
        df[col].max()
    ]

    edges = sorted(set(edges))

    df[col] = pd.cut(
        df[col],
        bins=edges,
        labels=labels[:len(edges)-1],
        duplicates="drop"
    )

    return df
```

```python
labels = ["not_popular", "below_avg", "average", "popular"]

df = categorize_col(df, "Vote_Average", labels)
```

```python
# Convert Release_Date directly to numeric
df["Release_Date"] = pd.to_numeric(df["Release_Date"], errors="coerce")

# Filter from 2015 onwards
df_2015 = df[df["Release_Date"] >= 2015]

# Count movies per year
year_counts = df_2015["Release_Date"].value_counts().sort_index()

# Pie chart
plt.pie(
    year_counts.values,
    labels=year_counts.index,
    autopct="%1.1f%%"
)

plt.title("Movies Released from 2015 Onwards")
plt.show()
```

```python
year_counts = df["Release_Date"].value_counts()

highest_year = year_counts.idxmax()
highest_count = year_counts.max()

lowest_year = year_counts.idxmin()
lowest_count = year_counts.min()

print("Highest:", highest_year, highest_count)
print("Lowest:", lowest_year, lowest_count)
```
